# Titanic Exploratory Data Analysis
This project performs Exploratory Data Analysis (EDA) on the Titanic dataset.

The main objectives are:
- Understand the dataset
- Identify and handle missing values
- Analyze survival patterns
- Compare survival based on gender and passenger class
- Explore passenger age distribution


## Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Loading the Dataset

```python
df = sns.load_dataset("titanic")
```

## Data Exploration

```python
print(df.head())
```

```python
print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

## Data Cleaning

```python
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])

df = df.drop(columns=["deck"])

print(df.isnull().sum())
```

## EDA Visualizations

```python
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
sns.histplot(df["age"], bins=20, kde=True)
plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

## Key Findings

```python
print("Total Passengers:", len(df))
print("Survived:", df["survived"].sum())
print("Not Survived:", len(df) - df["survived"].sum())

print("\nAverage Age:", df["age"].mean())
print("Average Fare:", df["fare"].mean())

print("\nSurvival Rate by Gender:")
print(df.groupby("sex")["survived"].mean())

print("\nSurvival Rate by Passenger Class:")
print(df.groupby("pclass")["survived"].mean())
```

## Conclusion

The Titanic dataset was analyzed using Exploratory Data Analysis and Data Visualization.

Key findings:
- Female passengers had a much higher survival rate than male passengers.
- First-class passengers had the highest survival rate.
- Third-class passengers had the lowest survival rate.
- Most passengers were young and middle-aged.
- Missing values were handled before analysis.

Overall, gender and passenger class had a strong relationship with survival.
