# Titanic Data Visualization

This project focuses on visualizing the Titanic dataset using charts and graphs.

The main objectives are:
- Understand the dataset
- Clean missing values
- Create meaningful visualizations
- Compare survival based on gender and passenger class
- Analyze age and fare distributions

## Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Loading the Dataset

```python
df = sns.load_dataset("titanic")
```

## Data Exploration

```python
print(df.head())
```

```python
print("Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())
```

## Data Cleaning

```python
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])

df = df.drop(columns=["deck"])

print(df.isnull().sum())
```

## Data Visualization

```python
sns.countplot(x="survived", data=df)
plt.title("Survival Count")
plt.show()
```

```python
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(x="pclass", hue="survived", data=df)
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()
```

```python
sns.histplot(df["age"], bins=20, kde=True)
plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
sns.histplot(df["fare"], bins=20, kde=True)
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Count")
plt.show()
```

## Conclusion

The Titanic dataset was analyzed using Exploratory Data Analysis and Data Visualization.

Key findings:
- Female passengers had a much higher survival rate than male passengers.
- First-class passengers had the highest survival rate.
- Third-class passengers had the lowest survival rate.
- Most passengers were young and middle-aged.
- Missing values were handled before analysis.

Overall, gender and passenger class had a strong relationship with survival.
